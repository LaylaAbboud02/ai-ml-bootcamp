```python
fig = plt.figure()

# Add a plotting area (Axes) to the figure.
# Format: [left, bottom, width, height]
# Values are fractions of the figure size, usually between 0 and 1.
# Example: [0.1, 0.1, 0.8, 0.8] means:
# start 10% from the left and bottom, and use 80% of the width and height.
ax = fig.add_axes([0.1, 0.1, 0.8, 0.8])

ax.plot(x, y)

plt.show()
```

## Matplotlib example workflow

```python
# 0. Import matplotlib and get it ready for plotting in jupyter notebook
import matplotlib.pyplot as plt

# 1. Prepare data
x = [1,2,3,4]
y = [11,22,33,44]

# 2. Setup plot
fig, ax = plt.subplots(figsize=(10,10)) # figsize is the size of the figure (as in width and height)

# 3. Plot data
ax.plot(x,y)

# 4. Customize plot
ax.set(title = "Simple PLot", xlabel = "x axis", ylabel = "y axis")

# 5. Display plot
plt.show()

# 6. Save plot
fig.savefig("images/sample-plot.png")
```

## Making figures with NumPy arrays
We're going to see:
* Line plot
* Scatter plot
* Bar plot
* Hist
* Subplots

```python
import numpy as np

# Create a NumPy array
x = np.linspace(0, 10, 100) # 100 evenly spaced numbers between 0 and 10
x[:100]
```

```python
# plot the array and create a line plot
fig, ax = plt.subplots()
ax.plot(x, x**2)
plt.show()
```

```python
# Use same data to make a scatter plot
fig, ax = plt.subplots()
ax.scatter(x, np.exp(x))
plt.show()
```

```python
# Another scatter plot
fix, ax = plt.subplots()
ax.scatter(x, np.sin(x)) # np.sin(x) is the sine of x
plt.show()
```

```python
# Make a bar plot from a dictionary
nut_butter_prices = {"Almonst butter": 10,
                     "Peanut butter": 8,
                     "Cashew butter": 12}   

fig, ax = plt.subplots()
ax.bar(
    nut_butter_prices.keys(),
    nut_butter_prices.values(),
    )
ax.set(title="Nut butter prices", ylabel="Price")
plt.show()
```

```python
fig, ax = plt.subplots()
ax.barh(
    y=nut_butter_prices.keys(),
    width=nut_butter_prices.values(),
    )
ax.set(title="Nut butter prices", xlabel="Price")
plt.show()
```

```python
# Make some data for histogram and plot it
x = np.random.randn(1000) 

fig, ax = plt.subplots()
ax.hist(x)
plt.show()
```

```python
# Subplpot option 1
fig, ((ax1,ax2), (ax3,ax4)) = plt.subplots(nrows=2, ncols=2, figsize=(10,5))
ax1.plot(x, x/2)
ax2.scatter(np.random.random(10), np.random.random(10))
ax3.bar(nut_butter_prices.keys(), nut_butter_prices.values())
ax4.hist(np.random.randn(1000))
plt.show()
```

```python
# Subplot option 2
fig, ax = plt.subplots(nrows=2, ncols=2, figsize=(10,5))
ax[0,0].plot(x, x/2)
ax[0,1].scatter(np.random.random(10), np.random.random(10))
ax[1,0].bar(nut_butter_prices.keys(), nut_butter_prices.values())
ax[1,1].hist(np.random.randn(1000))
plt.show()
```

## Plotting from pandas DataFrames

```python
import pandas as pd
```

```python
# Make a dataframe
car_sales = pd.read_csv("../sample_project/car-sales.csv")
car_sales
```

```python
# Create a time series
ts = pd.Series(np.random.randn(1000), 
               index=pd.date_range("1/1/2020", periods=1000)
               )

# Cumulative sum
ts = ts.cumsum()

# Plot the time series
ts.plot()
```

```python
car_sales
```

```python
# Convert the price column to a numeric type
car_sales["Price"] = car_sales["Price"].str.replace("[\$\,\.]", "", regex=True)
type(car_sales["Price"][0])
```

```python
# Remove the last two characters from the price column
car_sales["Price"] = car_sales["Price"].str[:-2]
car_sales
```

```python
car_sales["Sale Date"] = pd.date_range("1/1/2020", periods=len(car_sales))
car_sales
```

```python
car_sales["Price"] = car_sales["Price"].astype(int)
car_sales["Total Sales"] = car_sales["Price"].cumsum()
car_sales
```

```python
# Let's plot the total sales
fig, ax = plt.subplots()
ax.plot(car_sales["Sale Date"], car_sales["Total Sales"])
plt.show()
```

```python
# Another way to plot the total sales
car_sales.plot(x="Sale Date", y="Total Sales")
plt.show()
```

```python
car_sales.plot(x="Odometer (KM)", y="Price", kind="scatter")
plt.show()
```
