## Matplotlib example workflow

```python
# 0. Import matplotlib and get it ready for plotting in jupyter notebook
import matplotlib.pyplot as plt

# 1. Prepare data
x = [1,2,3,4]
y = [11,22,33,44]

# 2. Setup plot
fig, ax = plt.subplots(figsize=(10,10)) # figsize is the size of the figure (as in width and height)

# 3. Plot data
ax.plot(x,y)

# 4. Customize plot
ax.set(title = "Simple PLot", xlabel = "x axis", ylabel = "y axis")

# 5. Display plot
plt.show()

# 6. Save plot
fig.savefig("images/sample-plot.png")
```

## Making figures with NumPy arrays
We're going to see:
* Line plot
* Scatter plot
* Bar plot
* Hist
* Subplots

```python
import numpy as np

# Create a NumPy array
x = np.linspace(0, 10, 100) # 100 evenly spaced numbers between 0 and 10
x[:100]
```

```python
# plot the array and create a line plot
fig, ax = plt.subplots()
ax.plot(x, x**2)
plt.show()
```

```python
# Use same data to make a scatter plot
fig, ax = plt.subplots()
ax.scatter(x, np.exp(x))
plt.show()
```

```python
# Another scatter plot
fix, ax = plt.subplots()
ax.scatter(x, np.sin(x)) # np.sin(x) is the sine of x
plt.show()
```

```python
# Make a bar plot from a dictionary
nut_butter_prices = {"Almonst butter": 10,
                     "Peanut butter": 8,
                     "Cashew butter": 12}   

fig, ax = plt.subplots()
ax.bar(
    nut_butter_prices.keys(),
    nut_butter_prices.values(),
    )
ax.set(title="Nut butter prices", ylabel="Price")
plt.show()
```

```python
fig, ax = plt.subplots()
ax.barh(
    y=nut_butter_prices.keys(),
    width=nut_butter_prices.values(),
    )
ax.set(title="Nut butter prices", xlabel="Price")
plt.show()
```

```python
# Make some data for histogram and plot it
x = np.random.randn(1000) 

fig, ax = plt.subplots()
ax.hist(x)
plt.show()
```

```python
# Subplpot option 1
fig, ((ax1,ax2), (ax3,ax4)) = plt.subplots(nrows=2, ncols=2, figsize=(10,5))
ax1.plot(x, x/2)
ax2.scatter(np.random.random(10), np.random.random(10))
ax3.bar(nut_butter_prices.keys(), nut_butter_prices.values())
ax4.hist(np.random.randn(1000))
plt.show()
```

```python
# Subplot option 2
fig, ax = plt.subplots(nrows=2, ncols=2, figsize=(10,5))
ax[0,0].plot(x, x/2)
ax[0,1].scatter(np.random.random(10), np.random.random(10))
ax[1,0].bar(nut_butter_prices.keys(), nut_butter_prices.values())
ax[1,1].hist(np.random.randn(1000))
plt.show()
```

### Pyplot vs matplotlib object oriented method
* When plotting something quickly, okay to use the pyplot method
* When plotting something more advances, use OO

```python
over_50 = heart_disease[heart_disease["age"] > 50]
over_50
```

```python
# pyplot method
over_50.plot(kind='scatter', x='age', y='chol', c='target')
```

```python
# OO method
fix, ax = plt.subplots(figsize=(10,6))
over_50.plot(kind='scatter', x='age', y='chol', c='target', ax=ax)

ax.set_xlim([50,80])
```

```python
## OO method from scratch
fig, ax = plt.subplots(figsize=(10, 6))

# Plot the data
scatter = ax.scatter(x=over_50["age"], 
                     y = over_50["chol"], 
                     c=over_50["target"])

# Customize
ax.set(title = "Heart disease and cholesterol levels", 
       xlabel = "Age", 
       ylabel = "Cholesterol")

# Add a legent
ax.legend(*scatter.legend_elements(), title="Target")


# Add horizontal line
ax.axhline(over_50["chol"].mean(), linestyle="--")
```

```python
# subplot of chol, age, thalatch
fig, (ax0, ax1) = plt.subplots(nrows = 2, ncols = 1, figsize= (10,10), sharex=True)
fig.suptitle("Heart Disease Analysis", fontsize = 16, fontweight = "bold")

# Add data to ax0
scatter = ax0.scatter(x=over_50["age"], y=over_50["chol"], c=over_50["target"])

# Customize ax0
ax0.set(title="Heart disease and cholesterol levels", ylabel = "Cholesterol")

# Add a legent to ax0
ax0.legend(*scatter.legend_elements(), title="Target")

# Add a meanline
ax0.axhline(y=over_50["chol"].mean(), linestyle = "--")


# Add data to ax1
scatter = ax1.scatter(x=over_50["age"], y=over_50["thalach"], c=over_50["target"])

# Customize ax1
ax1.set(title="Heart disease and max heart rate", xlabel = "Age", ylabel = "Max Heart Rate")

# Add a legent to ax1
ax1.legend(*scatter.legend_elements(), title="Target")

# Add a meanline
ax1.axhline(y=over_50["thalach"].mean(), linestyle = "--")
```
