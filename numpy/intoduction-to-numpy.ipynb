## Manipulating & Comparing  Arrays

### Arithmetic

```python
array1 = np.array([1, 2, 3]) # Create a 1D array with elements from 1 to 3
array1
```

```python
ones = np.ones((3)) # Create a col array filled with ones
ones
```

```python
array1 + ones
```

```python
array1 - ones
```

```python
array2
```

```python
array1 * array2 # Element-wise multiplication of `array1` and `array2`. Since `array1` is 1D and `array2` is 2D, Numpy will broadcast `array1` across the rows of `array2`, resulting in a new 2D array where each element of `array1` multiplies the corresponding row of `array2`.
```

```python
array1/ones
```

```python
array2 / array1 # Element-wise division of `array2` by `array1`. Similar to multiplication, Numpy will broadcast `array1` across the rows of `array2`, resulting in a new 2D array where each element of `array2` is divided by the corresponding element of `array1`.
```

```python
array2 // array1 # Element-wise floor division of `array2` by `array1`. Similar to the multiplication operation, Numpy will broadcast `array1` across the rows of `array2`, resulting in a new 2D array where each element of `array2` is divided by the corresponding element of `array1`, and the result is floored to the nearest integer.
```

```python
array2 ** 2 # Element-wise exponentiation of `array2`. Each element of `array2` is raised to the power of 2, resulting in a new 2D array where each element is the square of the corresponding element in `array2`.
```

```python
array1 % 2 # Element-wise modulus operation on `array1`. This will return a new array where each element is the remainder of the corresponding element in `array1` divided by 2.
```

```python
np.mean(array2) # The `np.mean` function calculates the mean (average) of all elements in the array. For `array2`, it will return the average value of all its elements.
```

```python
np.max(array2) # The `np.max` function returns the maximum value from the array. For `array2`, it will return the largest element present in the array.
```

```python
np.std(array2) # The `np.std` function calculates the standard deviation of all elements in the array. For `array2`, it will return a measure of the amount of variation or dispersion of the elements in the array.
```

```python
np.var(array2) # The `np.var` function calculates the variance of all elements in the array. For `array2`, it will return a measure of how much the elements in the array deviate from the mean. Variance is the square of the standard deviation.
```

```python
# Standard deviation = square root of variance
np.sqrt(np.var(array2)) # The `np.sqrt` function calculates the square root of the input value. Here, it is used to calculate the square root of the variance of `array2`, which gives the standard deviation. This is equivalent to using `np.std(array2)`.
```
